```python
import requests

def get_weather(city):
    """Fetches current weather data for a given city."""
    response = requests.get(
        "https://api.openweathermap.org/data/2.5/weather",
        params={
            "q": city,
            "appid": OPENWEATHER_KEY,
            "units": "metric"
        },
        timeout=10
    )
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error fetching weather: {response.json().get('message', 'Unknown error')}")
        return None

def display_weather(data):
    """Displays current weather in a user-friendly format."""
    print(f"\n--- Current Weather in {data['name']}, {data['sys']['country']} ---")
    print(f"Temperature: {data['main']['temp']} °C (Feels like: {data['main']['feels_like']} °C)")
    print(f"Humidity: {data['main']['humidity']}%")
    print(f"Wind Speed: {data['wind']['speed']} m/s")
    print(f"Description: {data['weather'][0]['description'].capitalize()}")

def get_forecast(city):
    """Fetches and displays weather forecast for the city."""
    response = requests.get(
        "https://api.openweathermap.org/data/2.5/forecast",
        params={
            "q": city,
            "appid": OPENWEATHER_KEY,
            "units": "metric"
        },
        timeout=10
    )
    if response.status_code == 200:
        data = response.json()
        print(f"\n--- Forecast for {data['city']['name']} (Next 24 Hours) ---")
        # Loop through forecast items in 3-hour intervals (showing the next 8 items for 24 hours)
        for item in data['list'][:8]:
            time = item['dt_txt']
            temp = item['main']['temp']
            desc = item['weather'][0]['description']
            print(f"Time: {time} | Temp: {temp} °C | {desc.capitalize()}")
    else:
        print("Could not fetch forecast details.")

while True:
    city = input("\nEnter city name (blank to exit): ").strip()

    if not city:
        print("Thank you for using Weather App!")
        break

    weather_data = get_weather(city)

    if weather_data:
        display_weather(weather_data)
        get_forecast(city)
```

# Weather App with Live OpenWeather API

## Project Description
The Weather App is a Python application that retrieves real-time weather information using the OpenWeather API. Users can enter a city name to view temperature, humidity, wind speed, weather conditions and a five-day forecast.

## Technologies Used
- Python
- Google Colab
- OpenWeather API
- Requests Library
- JSON

## Features
- Search weather by city name
- Display live temperature in Celsius
- Display humidity percentage
- Display wind speed
- Show weather descriptions
- Five-day weather forecast
- Handle invalid city names
- Handle API errors

## API Integration
The application connects to the OpenWeather REST API using Python Requests. The API response is processed in JSON format.

## API Key Security
The API key is entered securely using Python getpass. It is not displayed in the notebook output or included in the public repository.

## Conclusion
This project demonstrates Python programming, REST API integration, JSON parsing and exception handling. It provides real-time weather information through a simple command-line application.
